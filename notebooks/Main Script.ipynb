```python
# Cell 1: setup helper (run once)
import copy
import yaml
from aimo3.scripts.helpers import load_yaml, get_config_path
from aimo3.scripts.sft import train_model

def run_sft_once(base_cfg_name, exp_name):
    cfg = copy.deepcopy(load_yaml(get_config_path(base_cfg_name)))
    cfg["run"]["experiment_name"] = exp_name
    tmp_cfg = f"/content/{base_cfg_name.replace('.yaml','')}_{exp_name}.yaml"
    with open(tmp_cfg, "w") as f:
        yaml.safe_dump(cfg, f, sort_keys=False)
    print(f"Running {base_cfg_name} on {exp_name}")
    train_model(tmp_cfg)
```

```python
# Cell 2: DeepSeekMath exp1
run_sft_once("sft_deepseekmath.yaml", "exp1")
```

```python
# Cell 3: DeepSeekMath exp2
run_sft_once("sft_deepseekmath.yaml", "exp2")
```

```python
# Cell 3: DeepSeekMath exp2
run_sft_once("sft_deepseekmath.yaml", "exp3")
```
