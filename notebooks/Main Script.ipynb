```python
from datasets import load_from_disk
import os

base = "/content/drive/MyDrive/Math Olympiad Competition/Experimentation/processed_splits/omr_aimo3/splits"
experiments = ["exp1", "exp2", "exp3"]
n = "50000"

summary = {}
for exp in experiments:
    p = os.path.join(base, exp, n)
    if not os.path.exists(p):
        raise FileNotFoundError(f"Missing split folder: {p}")

    ds = load_from_disk(p)
    train_n = len(ds["train"])
    val_n = len(ds["val"])
    val2_n = len(ds["val2"])
    test_n = len(ds["test"])
    total = train_n + val_n + val2_n + test_n

    summary[exp] = {
        "train": train_n,
        "val": val_n,
        "val2": val2_n,
        "test": test_n,
        "total": total,
    }

print("Per-experiment sizes:")
for exp, s in summary.items():
    print(exp, s)

totals = {s["total"] for s in summary.values()}
if totals != {50000}:
    raise ValueError(f"Totals are not all 50,000: {summary}")

# Optional strict check: exact same split sizes for all experiments
split_shape_sets = {
    (s["train"], s["val"], s["val2"], s["test"])
    for s in summary.values()
}
if len(split_shape_sets) != 1:
    print("Totals are equal, but split buckets differ slightly due to grouped split logic.")
else:
    print("Exact split sizes match across all experiments.")

print("Check complete.")
```

```python
# Cell 1: setup helper (run once)
import copy
import yaml
from aimo3.scripts.helpers import load_yaml, get_config_path
from aimo3.scripts.sft import train_model

def run_sft_once(base_cfg_name, exp_name):
    cfg = copy.deepcopy(load_yaml(get_config_path(base_cfg_name)))
    cfg["run"]["experiment_name"] = exp_name
    tmp_cfg = f"/content/{base_cfg_name.replace('.yaml','')}_{exp_name}.yaml"
    with open(tmp_cfg, "w") as f:
        yaml.safe_dump(cfg, f, sort_keys=False)
    print(f"Running {base_cfg_name} on {exp_name}")
    train_model(tmp_cfg)
```

# sft.py

- Loads experiment, data, and model settings from YAML, then resolves:
  - Which experiment/model to run
  - Where prepared splits live
  - Where SFT outputs/checkpoints should be saved
- Converts each example into supervised fine-tuning text with this structure:
  - System prompt, Problem, Solution (reasoning), FINAL_ANSWER
- Loads precomputed train/val/test splits from disk and reports split sizes.
- Formats splits for training
- Builds tokenizer settings for causal LM training:
  - Ensures a valid pad token, Uses right-padding
- Enables 4-bit quantization (BitsAndBytes) and prepares model for k-bit training.
- Builds LoRA PEFT config from YAML
- Builds TRL SFTTrainer config from YAML hyperparameters
- Saves reproducibility metadata to output folder
- Runs end-to-end training with optional resume-from-checkpoint support.
- Saves final trained model and tokenizer to the configured output directory.
