# SFT Dataset Preview (Before Training)

This notebook shows one real dataset row right before SFT formatting and the exact final text fed into SFT.

```python
import os
from datasets import load_from_disk

def convert_to_prompt_format(example, system_prompt):
    problem = str(example["problem"]).strip()
    reasoning = str(example["generated_solution"]).strip()
    answer = str(example["expected_answer"]).strip()

    text = (
        system_prompt
        + "\n\nProblem:\n"
        + problem
        + "\n\nSolution:\n"
        + reasoning
        + "\nFINAL_ANSWER: "
        + answer
    )

    return {
        "text": text,
        "expected_answer": answer,
    }
```

```python
# Set these to your actual values
prepared_data_path = "/path/to/processed_splits/splits/exp2/100"
system_prompt = """You are a mathematical problem solver.
Solve step-by-step.
At the end, output a line exactly of the form:
FINAL_ANSWER: <integer>
Where <integer> is an integer between 0 and 99999."""

prepared = load_from_disk(prepared_data_path)
print("Available splits:", list(prepared.keys()))
for name in prepared.keys():
    print(f"{name}: {len(prepared[name])}")
```

```python
split_name = "train"
example_index = 0

ds = prepared[split_name]
ex = ds[example_index]

print("RAW problem:\n")
print(str(ex["problem"]).strip())
print("\nRAW generated_solution:\n")
print(str(ex["generated_solution"]).strip())
print("\nRAW expected_answer:\n")
print(str(ex["expected_answer"]).strip())
```

```python
formatted = convert_to_prompt_format(ex, system_prompt)

print("FORMATTED text fed to SFT:\n")
print(formatted["text"])
print("\nEXPECTED ANSWER FIELD:", formatted["expected_answer"])
```

```python
# Optional quick preview of the first 3 examples
for i in range(min(3, len(ds))):
    row = ds[i]
    p = str(row["problem"]).strip().replace("\n", " ")
    a = str(row["expected_answer"]).strip()
    print(f"[{i}] problem: {p[:120]}...")
    print(f"    expected_answer: {a}")
```
